```python
import numpy as np

# Number of junctions
n_junctions = 4

# Traffic queue at each junction
queue = np.array([40, 70, 30, 90])

# PSO parameters
n_particles = 20
iterations = 50
w = 0.7
c1 = 1.5
c2 = 1.5

# Green light limits
min_green = 10
max_green = 60


# Fitness function
def fitness(green_times):
    # More green time reduces the queue
    remaining_queue = np.maximum(queue - green_times, 0)

    # Total remaining traffic is the cost to minimize
    return np.sum(remaining_queue)


# Initialize particles randomly
positions = np.random.uniform(
    min_green, max_green,
    (n_particles, n_junctions)
)

velocities = np.random.uniform(
    -5, 5,
    (n_particles, n_junctions)
)

# Personal best
pbest_positions = positions.copy()
pbest_scores = np.array([fitness(p) for p in positions])

# Global best
gbest_index = np.argmin(pbest_scores)
gbest_position = pbest_positions[gbest_index].copy()
gbest_score = pbest_scores[gbest_index]


# PSO iterations
for iteration in range(iterations):

    for i in range(n_particles):

        r1 = np.random.rand(n_junctions)
        r2 = np.random.rand(n_junctions)

        # Update velocity
        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pbest_positions[i] - positions[i])
            + c2 * r2 * (gbest_position - positions[i])
        )

        # Update position
        positions[i] += velocities[i]

        # Keep green time between 10 and 60 seconds
        positions[i] = np.clip(
            positions[i],
            min_green,
            max_green
        )

        # Calculate fitness
        score = fitness(positions[i])

        # Update personal best
        if score < pbest_scores[i]:
            pbest_scores[i] = score
            pbest_positions[i] = positions[i].copy()

            # Update global best
            if score < gbest_score:
                gbest_score = score
                gbest_position = positions[i].copy()


# Output result
print("Optimal Green Light Times:")
for j in range(n_junctions):
    print(
        f"Junction {j + 1}: "
        f"{gbest_position[j]:.2f} seconds"
    )

print("\nMinimum Traffic Queue:", gbest_score)
```
