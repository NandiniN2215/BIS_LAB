```python
import random

classes=[
    ("Math","F1"),
    ("Physics","F2"),
    ("Chemistry","F3"),
    ("English","F1"),
    ("Computer","F4")
]

rooms=["R1","R2"]
slots=["9AM","10AM","11AM","12PM"]

POPULATION_SIZE=20
GENERATIONS=100
MUTATION_RATE=0.1

def create_chromosome():
    return [
        (random.choice(slots),random.choice(rooms))
        for _ in range(len(classes))
    ]

def create_population():
    return [create_chromosome() for _ in range(POPULATION_SIZE)]

def count_clashes(chromosome):
    clashes=0

    for i in range(len(classes)):
        for j in range(i+1,len(classes)):

            slot1,room1=chromosome[i]
            slot2,room2=chromosome[j]

            faculty1=classes[i][1]
            faculty2=classes[j][1]

            # Same room at same time
            if slot1==slot2 and room1==room2:
                clashes+=1

            # Same faculty at same time
            if slot1==slot2 and faculty1==faculty2:
                clashes+=1

    return clashes

def fitness(chromosome):
    clashes=count_clashes(chromosome)
    return 1/(1+clashes)

def selection(population):
    population.sort(key=fitness,reverse=True)
    return population[:2]

def crossover(parent1,parent2):
    point=random.randint(1,len(classes)-1)
    child=parent1[:point]+parent2[point:]
    return child

def mutation(chromosome):
    for i in range(len(chromosome)):

        if random.random()<MUTATION_RATE:
            slot=random.choice(slots)
            room=random.choice(rooms)
            chromosome[i]=(slot,room)

    return chromosome

population=create_population()

for generation in range(GENERATIONS):

    parents=selection(population)
    new_population=parents.copy()

    while len(new_population)<POPULATION_SIZE:

        parent1=random.choice(parents)
        parent2=random.choice(parents)

        child=crossover(parent1,parent2)
        child=mutation(child)

        new_population.append(child)

    population=new_population

best=max(population,key=fitness)

print("Final Timetable")
print("----------------")

for i in range(len(classes)):

    subject,faculty=classes[i]
    slot,room=best[i]

    print(
        subject,
        "| Faculty:",faculty,
        "| Slot:",slot,
        "| Room:",room
    )

print("\nNumber of clashes:",count_clashes(best))
print("Fitness:",fitness(best))
```
